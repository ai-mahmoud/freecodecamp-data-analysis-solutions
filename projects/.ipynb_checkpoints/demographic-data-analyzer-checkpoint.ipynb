```python
import pandas as pd
```

```python
df = pd.read_csv("adult.data.csv")
df['race'].value_counts()
```

```python
df[df['sex'] == 'Male']['age'].mean()
```

```python
df[df['education'] == 'Bachelors'].count() / df.count()
```

```python
df[(df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])) & (df['salary'].isin(['<=50K']))]
```

```python
df[~(df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])) & (df['salary'].isin(['<=50K']))]
```

```python
df['hours-per-week'].min()
```

```python
df[(df['hours-per-week'] == df['hours-per-week'].min()) & (df['salary'].isin(['<=50K']))]
```

```python
counts = df['native-country'].value_counts()
```

```python
country = counts.idxmax()
```

```python
percentage = counts.max() / counts.sum() * 100
```

```python
print(f'country is {country} and percentage is {percentage:.2f}%')
```
