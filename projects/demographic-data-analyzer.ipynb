```python
import pandas as pd
```

```python
df = pd.read_csv("data/adult.data.csv")
df['race'].value_counts()
```

```python
round(df[df['sex'] == 'Male']['age'].mean(), 1)
```

```python
round((len(df[df['education'] == 'Bachelors']) / len(df)) * 100, 1)
```

```python
higher_education =  df[df['education'].isin(['Bachelors', 'Masters', 'Doctorate'])]
lower_education = df[~(df['education'].isin(['Bachelors', 'Masters', 'Doctorate']))]
```

```python
round((len(higher_education[higher_education['salary'] == '>50K']) / len(higher_education)) * 100, 1)
```

```python
round((len(lower_education[lower_education['salary'] == '>50K']) / len(lower_education)) * 100, 1)
```

```python
min_work_hours = df['hours-per-week'].min()
min_work_hours
```

```python
num_min_workers = len(df[df['hours-per-week'] == min_work_hours])

rich_percentage = round(
        (len(df[(df['hours-per-week'] == min_work_hours) &
        (df['salary'] == '>50K')]) / num_min_workers) * 100, 1)
rich_percentage
```

```python
country_earnings = df[df['salary'] == '>50K']['native-country'].value_counts()
country_totals = df['native-country'].value_counts()
country_percentages = (country_earnings / country_totals * 100).dropna()
highest_earning_country = country_percentages.idxmax()

highest_earning_country
```

```python
highest_earning_country_percentage = round(country_percentages.max(), 1)
highest_earning_country_percentage
```

```python
top_IN_occupation = df[(df['native-country'] == 'India') & (df['salary'] == '>50K')]['occupation'].value_counts().idxmax()
top_IN_occupation
```
