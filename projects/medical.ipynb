```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/medical_examination.csv")
df
```

```python
df["overweight"] = (df["weight"] / (df["height"]/100)**2 > 25).astype(np.int64)
df
```

```python
df['gluc'] = np.where( df['gluc'] == 1 , 0, 1)
df['cholesterol'] = np.where( df['cholesterol'] == 1 , 0, 1)
df
```

```python
df_cat = pd.melt(df, id_vars=['cardio'], 
                     value_vars=['cholesterol', 'gluc', 'smoke', 'alco', 'active', 'overweight'])
```

```python
df_cat = df_cat.groupby(['cardio', 'variable', 'value'])['value'].count().reset_index(name='total')
```

```python
g = sns.catplot(
        data=df_cat, 
        x='variable',
        y='total',
        hue='value', 
        col='cardio', 
        kind='bar')
g
```

```python
fig = g.fig
fig
```

```python
df_heat = df[
(df['ap_lo'] <= df['ap_hi']) &
(df['height'] >= df['height'].quantile(0.025)) &
(df['height'] <= df['height'].quantile(0.975)) &
(df['weight'] >= df['weight'].quantile(0.025)) &
(df['weight'] <= df['weight'].quantile(0.975))
]
df_heat
```

```python
corr = df_heat.corr()
corr
```

```python
mask = np.triu(np.ones_like(corr, dtype=bool))
```

```python
fig, ax = plt.subplots(figsize=(12, 12))
sns.heatmap(corr, mask=mask, annot=True, fmt='.1f', linewidths=1, square=True, ax=ax)
plt.show()
```
